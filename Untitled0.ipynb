```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import silhouette_score, accuracy_score, f1_score
import joblib

# 1. Generate realistic customer dataset
np.random.seed(42)
n = 1000
df = pd.DataFrame({
    'CustomerID': range(1001, 1001 + n),
    'Gender': np.random.choice(['Male', 'Female'], n),
    'Age': np.random.randint(18, 70, n),
    'Annual_Income_Lakhs': np.round(np.random.uniform(2.5, 25.0, n), 2),
    'Spending_Score': np.random.randint(1, 100, n),
    'Purchase_Frequency': np.random.randint(1, 30, n),
})

response_prob = (
    0.25 * (df['Spending_Score'] / 100) +
    0.25 * (df['Purchase_Frequency'] / 30) +
    0.20 * (df['Annual_Income_Lakhs'] / 25) +
    np.random.uniform(0, 0.3, n)
)
df['Campaign_Response'] = (response_prob > 0.52).astype(int)
df.to_csv('raw_customer_data.csv', index=False)

# 2. Data Preprocessing & Cleaning
df_clean = df.dropna().copy()
df_clean['Gender_Code'] = df_clean['Gender'].map({'Male': 0, 'Female': 1})

features = ['Age', 'Annual_Income_Lakhs', 'Spending_Score', 'Purchase_Frequency', 'Gender_Code']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_clean[features])

# 3. ML Model 1: Customer Segmentation (K-Means)
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df_clean['Segment_ID'] = kmeans.fit_predict(X_scaled)
sil_score = silhouette_score(X_scaled, df_clean['Segment_ID'])

cluster_summary = df_clean.groupby('Segment_ID')[['Annual_Income_Lakhs', 'Spending_Score']].mean()
segment_names = {}
for cid, row in cluster_summary.iterrows():
    if row['Annual_Income_Lakhs'] >= 13 and row['Spending_Score'] >= 50:
        segment_names[cid] = "VIP Big Spenders"
    elif row['Annual_Income_Lakhs'] >= 13 and row['Spending_Score'] < 50:
        segment_names[cid] = "Affluent Careful Savers"
    elif row['Annual_Income_Lakhs'] < 13 and row['Spending_Score'] >= 50:
        segment_names[cid] = "Young Impulse Buyers"
    else:
        segment_names[cid] = "Budget Value Shoppers"

df_clean['Segment_Name'] = df_clean['Segment_ID'].map(segment_names)

# 4. ML Model 2: Campaign Response Prediction (Random Forest)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, df_clean['Campaign_Response'], test_size=0.2, random_state=42)
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)
y_pred = rf_model.predict(X_test)

acc = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

df_clean.to_csv('cleaned_data.csv', index=False)
joblib.dump({'scaler': scaler, 'kmeans': kmeans, 'rf_model': rf_model, 'segment_names': segment_names, 'features': features}, 'models.pkl')

print(f"K-Means Silhouette Score : {sil_score:.3f}")
print(f"Campaign Model Accuracy  : {acc*100:.2f}%")
print(f"Campaign Model F1-Score  : {f1:.3f}")
```

```python
import pandas as pd
import plotly.express as px
import joblib
import ipywidgets as widgets
from IPython.display import display, clear_output

# Load your cleaned dataset and trained ML models
df = pd.read_csv('cleaned_data.csv')
bundle = joblib.load('models.pkl')
scaler, kmeans, rf_model = bundle['scaler'], bundle['kmeans'], bundle['rf_model']
segment_names, features = bundle['segment_names'], bundle['features']

recommendations = {
    "VIP Big Spenders": "Offer VIP loyalty tier, early access to new launches, and personal concierge rewards.",
    "Affluent Careful Savers": "Send high-value bundle discounts, extended warranties, and quality-focused emails.",
    "Young Impulse Buyers": "Target with flash-sale push notifications, limited-time coupons, and social media ads.",
    "Budget Value Shoppers": "Promote cashback offers, festival mega-sales, and essential combo packs."
}

# 1. Interactive Scatter Plot: Customer Segments
fig1 = px.scatter(
    df, x="Annual_Income_Lakhs", y="Spending_Score", color="Segment_Name",
    title="1. Customer Segments: Annual Income (₹ Lakhs) vs. Spending Score",
    hover_data=["Age", "Purchase_Frequency"]
)
fig1.show()

# 2. Interactive Bar Chart: Campaign Response Rate by Segment
resp = df.groupby("Segment_Name")["Campaign_Response"].mean().reset_index()
resp["Response_Pct"] = (resp["Campaign_Response"] * 100).round(1)
fig2 = px.bar(
    resp, x="Segment_Name", y="Response_Pct", color="Segment_Name", text="Response_Pct",
    title="2. Predicted Campaign Response Rate by Segment (%)"
)
fig2.show()

# 3. Segment Profile & Marketing Strategy Table
profile = df.groupby("Segment_Name").agg(
    Customers=("CustomerID", "count"),
    Avg_Age=("Age", "mean"),
    Avg_Income_Lakhs=("Annual_Income_Lakhs", "mean"),
    Avg_Spending=("Spending_Score", "mean"),
    Avg_Frequency=("Purchase_Frequency", "mean"),
    Response_Rate_Pct=("Campaign_Response", lambda x: x.mean() * 100)
).round(1).reset_index()
profile["Recommended_Campaign"] = profile["Segment_Name"].map(
    lambda s: recommendations.get(s, "Send personalized seasonal discount coupon.")
)
print("\n=== 3. SEGMENT PROFILES & RECOMMENDED MARKETING ACTIONS ===")
display(profile)

# 4. Live Interactive Customer Predictor (Sliders right inside Colab!)
print("\n=== 4. LIVE CUSTOMER SEGMENT & CAMPAIGN PREDICTOR ===")
age_w = widgets.IntSlider(value=32, min=18, max=70, description='Age:')
inc_w = widgets.FloatSlider(value=14.5, min=2.5, max=25.0, step=0.5, description='Income (₹L):')
spd_w = widgets.IntSlider(value=75, min=1, max=100, description='Spending:')
frq_w = widgets.IntSlider(value=15, min=1, max=30, description='Frequency:')
gen_w = widgets.Dropdown(options=['Male', 'Female'], value='Male', description='Gender:')
btn = widgets.Button(description='Analyze Customer', button_style='success')
out = widgets.Output()

def on_click(b):
    with out:
        clear_output()
        gcode = 0 if gen_w.value == 'Male' else 1
        inp = pd.DataFrame([[age_w.value, inc_w.value, spd_w.value, frq_w.value, gcode]], columns=features)
        inp_scaled = scaler.transform(inp)
        seg_id = kmeans.predict(inp_scaled)[0]
        seg_name = segment_names.get(seg_id, f"Segment {seg_id}")
        prob = rf_model.predict_proba(inp_scaled)[0][1] * 100
        offer = recommendations.get(seg_name, "Send personalized seasonal discount coupon.")
        print(f"Assigned Segment            : {seg_name}")
        print(f"Campaign Response Likelihood: {prob:.1f}%")
        print(f"Recommended Action          : {offer}")

btn.on_click(on_click)
display(widgets.VBox([age_w, inc_w, spd_w, frq_w, gen_w, btn, out]))
on_click(None)
```
